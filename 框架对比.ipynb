## 2. 选型结论

| 你的核心诉求 | 首选 | 原因 |
|---|---|---|
| 文档/知识库问答 | **LlamaIndex** | Index、Retriever、QueryEngine 是核心抽象 |
| 审批流、固定步骤、可暂停恢复 | **LangGraph** | StateGraph、checkpoint、interrupt 最直接 |
| 多角色研究/审核/写作 | **CrewAI** | Agent/Task/Crew 的角色模型最自然 |
| 一个助手调用大量业务工具 | **LangChain** | Tool、Agent、Middleware 和集成生态最完整 |
| OpenAI 生态的分流/专家接管/护栏 | **OpenAI Agents SDK** | Handoff、Guardrail、Tracing 内置且 API 轻量 |

详细说明见 `docs/framework-selection.md`。


每个特性的原理和对应代码位置见 `docs/framework-feature-map.md`。
